```python
import dask
import dask.dataframe as dd
from dask.diagnostics import ProgressBar
import vaex

# Define the path to your original dataset (must be in Parquet format)
INPUT_PARQUET_PATH = "D:/jn/Amazon_cs_proj/Issue_categor_proj_Phase1/final_phase_ready_vaex/*.parquet"

# Define output folder for processed parquet files
OUTPUT_PARQUET_DIR = "processed_data/"

# Define your actual multi-label columns (True/False format)
LABEL_COLS = [
    'refund_and_returns',
    'delivery_delay',
    'product_quality',
    'order_issue',
    'packaging_handling',
    'long_wait_times',
    'technical_support',
    'warranty_guarantee',
    'fake_counterfeit',
    'rare_issues',
    'minor_issues'
]

# Step 1: Load using Dask
ddf = dd.read_parquet(INPUT_PARQUET_PATH)

# Step 2: Convert boolean labels to int (True → 1, False → 0)
ddf[LABEL_COLS] = ddf[LABEL_COLS].astype("int8")

# Step 3: Write processed data to Parquet using Dask with a progress bar
with ProgressBar():
    ddf.to_parquet(OUTPUT_PARQUET_DIR, write_index=False, engine='pyarrow', overwrite=True)

# Step 4: Load into Vaex from saved Parquet
vdf = vaex.open(f"{OUTPUT_PARQUET_DIR}/*.parquet")

# FOR step5 show it :
print(vdf.head(2))
print(vdf.shape)
```
